# 01 — Explore Source Data

Explore the source datasets defined in `configs/datasets.yaml`.
- Load manifests and inspect entries
- Check audio quality (sample rate, channels, duration distribution)
- Speaker distribution
- Transcript length distribution

```python
import json
import os
import yaml
import pandas as pd
import matplotlib.pyplot as plt
import soundfile as sf
from collections import Counter

with open('../configs/datasets.yaml') as f:
    config = yaml.safe_load(f)

print(f"Datasets: {list(config['datasets'].keys())}")
```

```python
# Load and inspect a manifest
ds_key = list(config['datasets'].keys())[0]
ds = config['datasets'][ds_key]
manifest_path = ds['manifest']
print(f"Dataset: {ds['name']}")
print(f"Manifest: {manifest_path}")

if os.path.isfile(manifest_path):
    entries = []
    with open(manifest_path) as f:
        for line in f:
            entries.append(json.loads(line.strip()))
    df = pd.DataFrame(entries)
    print(f"Entries: {len(df)}")
    display(df.head())
    display(df.describe())
else:
    print(f"Manifest not found: {manifest_path}")
```
