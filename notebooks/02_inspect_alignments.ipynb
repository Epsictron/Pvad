# 02 — Inspect Alignments

Compare raw NFA CTM vs refined (VAD-corrected) CTM.
- Visualize word boundaries on waveforms
- Compute boundary shift statistics
- Identify problematic alignments

```python
import json
import os
import numpy as np
import matplotlib.pyplot as plt
import soundfile as sf
import yaml

with open('../configs/datasets.yaml') as f:
    config = yaml.safe_load(f)

def load_ctm(ctm_path):
    words = []
    with open(ctm_path) as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 5:
                words.append({'start': float(parts[2]), 'duration': float(parts[3]), 'word': parts[4]})
    return words

print('Ready to inspect alignments.')
```

```python
# Compare raw vs refined CTM for a sample utterance
ds_path = 'hindi_conv'  # Change as needed
utt_id = 'utt_000000'

raw_ctm_path = f'/data/nfa_output/{ds_path}/ctm/words/{utt_id}.ctm'
refined_ctm_path = f'/data/refined_ctm/{ds_path}/{utt_id}.ctm'

if os.path.isfile(raw_ctm_path) and os.path.isfile(refined_ctm_path):
    raw = load_ctm(raw_ctm_path)
    refined = load_ctm(refined_ctm_path)
    print(f'Raw words: {len(raw)}, Refined words: {len(refined)}')
    
    shifts = [abs(r['start'] - ref['start']) for r, ref in zip(raw, refined)]
    print(f'Mean start shift: {np.mean(shifts):.4f}s')
    print(f'Max start shift: {np.max(shifts):.4f}s')
else:
    print('CTM files not found. Run PATH B scripts first.')
```
